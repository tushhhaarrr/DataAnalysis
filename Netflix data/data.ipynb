```python
def categorize(df,col,labels):
 edges = [
  df[col].describe()['min'],
  df[col].describe()['25%'],
  df[col].describe()['50%'],
  df[col].describe()['75%'],
  df[col].describe()['max']
 ]
 df[col]=pd.cut(df[col], edges ,labels=labels,duplicates='drop')
 return df
```

```python
labels=['Not_popular','Below_avg','average','popular']
categorize(df,'Vote_Average',labels)
df['Vote_Average'].unique()
```
